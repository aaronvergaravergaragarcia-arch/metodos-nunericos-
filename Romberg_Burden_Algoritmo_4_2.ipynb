
Integral: $I = \int_a^b f(x) dx$
Ejemplo 2 pág 160: $f(x)=sin(x)$ en $[0, \pi]$, $\int_0^{\pi} sin x dx = 2$

Fórmulas:
$R_{1,1}=h/2[f(a)+f(b)]$
$R_{i,j}=R_{i,j-1}+(R_{i,j-1}-R_{i-1,j-1})/(4^{j-1}-1)$

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

def romberg_burden_42(f, a, b, n):
    R = np.zeros((2, n))

    h = b - a
    R[0,0] = h/2 * (f(a) + f(b))
    print(f"Paso 1: h={h}, R11={R[0,0]:.8f}")

    print(f"Paso 2: SALIDA R11={R[0,0]:.8f}")

    for i in range(2, n+1):
        suma = 0
        for k in range(1, 2**(i-2)+1):
            suma += f(a + (k-0.5)*h)
        R[1,0] = 0.5 * (R[0,0] + h*suma)

        for j in range(2, i+1):
            R[1,j-1] = R[1,j-2] + (R[1,j-2]-R[0,j-2])/(4**(j-1)-1)

        print(f"i={i}: ", end="")
        for j in range(1, i+1):
            print(f"R{i},{j}={R[1,j-1]:.8f} ", end="")
        print()

        h = h/2
        for j in range(1, i+1):
            R[0,j-1] = R[1,j-1]

    return R[0,n-1]

def f(x): return np.sin(x)

sol = romberg_burden_42(f, 0, np.pi, 6)
print(f"\n SOLUCION FINAL R6,6 = {sol:.8f} (libro 2.00000000)")
```
